# Web Scraping Using Selenium

## Introduction

Web scraping is the process of automatically collecting data from websites using software or scripts. Instead of manually visiting web pages and copying information, a web scraping program can access web pages, identify the required elements, extract the relevant data, and store or process it for further use.

Web scraping is widely used for **data collection, research, market analysis, price monitoring, content aggregation, automation, and data analysis**. It helps reduce manual effort, saves time, and makes it possible to collect large amounts of publicly available information efficiently.

## Why Web Scraping is Important

Websites contain a large amount of useful information, but collecting this information manually can be time-consuming and repetitive. Web scraping automates this process and allows data to be collected in a structured and consistent way.

Some common applications of web scraping include:

* Collecting product names, prices, and details from e-commerce websites
* Extracting news articles and headlines
* Gathering job listings and employment information
* Collecting public reviews and ratings
* Monitoring changes in prices or website content
* Collecting data for research and analysis
* Automating repetitive browser-based tasks

## Tool Used: Selenium

For this project, **Selenium with Python** is used for web scraping and browser automation.

Selenium is an open-source tool that allows Python programs to control web browsers such as Google Chrome. Unlike simple HTTP-based scraping techniques, Selenium can interact with websites in a way similar to a real user. It can open web pages, click buttons, enter text, select options, scroll pages, and extract information from dynamically loaded content.

This makes Selenium particularly useful for websites where the required data is generated or loaded using **JavaScript** after the initial page loads.

### Main Selenium Features Used

* **Browser Automation** – Opens and controls the web browser programmatically.
* **Element Location** – Finds web elements using ID, Name, Class Name, CSS Selector, XPath, and other locator strategies.
* **Data Extraction** – Retrieves text, attributes, links, and other information from web elements.
* **User Interaction** – Allows automation of actions such as clicking, typing, scrolling, and selecting options.
* **Dynamic Content Handling** – Helps access content that is loaded dynamically through JavaScript.

## Basic Selenium Workflow

The general workflow followed in a Selenium-based web scraping project is:

1. Install Selenium and configure the Python environment.
2. Import the required Selenium modules.
3. Initialize the web browser.
4. Navigate to the target website.
5. Locate the required web elements.
6. Extract the required information.
7. Process or store the collected data.
8. Close the browser session.

````python
# Import the required Selenium modules
from selenium import webdriver
from selenium.webdriver.common.by import By
````

````python
'''
An overview of how data are stored in table format.
----------------------------------------------------

<table>
    <thead>
        <tr>
            <th>Heading 1</th>
            <th>Heading 2</th>
        </tr>
    </thead>

    <tbody>
        <tr>
            <td>Data 1</td>
            <td>Data 1</td>
        </tr>
    </tbody>
</table>
'''


'''
NOTE: Selenium only supports Chrome and Edge web browser.
'''
````

````python
# Website using
website = "https://merolagani.com/LatestMarket.aspx"
````

````python
# Initialize the web browser -> Chrome Driver
driver = webdriver.Chrome()
````

````python
# Navigate to the target website
driver.get(website)
````

````python
# Locate the required web elements
table_body = driver.find_element(By.TAG_NAME, 'tbody')
rows = table_body.find_elements(By.TAG_NAME, 'tr')
````

````python
# Arrange Data
for row in rows:
    # columns
    columns = row.find_elements(By.TAG_NAME, 'td')
    # column data
    for column in columns:
        # data are stored into objects. Converting into text
        print(column.text)
````

#### NOTE:
    - The Session period for fetching data is very short. So it might throw an error if we run the code taking long time interval.

### Process or store the collected data

````python
# Columns list to place datas.
symbol_col = []
ltp_col = []
change_col = []
open_col = []
high_col = []
low_col = []
qty_col = []
pclose_col = []
diff_col = []
````

````python
# Extract the required information
# Process or store the collected data
for row in rows:
    columns = row.find_elements(By.TAG_NAME, 'td')
    symbol_col.append(columns[0].text)
    ltp_col.append(columns[1].text)
    change_col.append(columns[2].text)
    open_col.append(columns[3].text)
    high_col.append(columns[4].text)
    low_col.append(columns[5].text)
    qty_col.append(columns[6].text)
    pclose_col.append(columns[7].text)
    diff_col.append(columns[8].text)
print('Data Stored Successfully.')
````

````python
# Check data stored into column or not
symbol_col
````

````python
# Creating data as a Dictionary
data = {
    "Symbol": symbol_col,
    "LTP": ltp_col,
    "Change": change_col,
    "Open": open_col,
    "High": high_col,
    "Low":  low_col,
    "Qty":  qty_col,
    "pClose": pclose_col,
    "Diff": diff_col
}
````

````python
# Close the browser session
driver.quit()
````

````python
# Check Dictionary store data or not
data
````
