## Data Cleaning Steps:
    1. Remove duplicate row data
        - check duplicate data --> duplicated()
        - remove duplicate data --> drop_duplicates()

    2. Drop Data
        - If columns contain at least 70-75% Null values then Drop that column.
        - Drop Column --> drop() --> pass column name to drop entire column or pass row index to drop entire row
        - dropna() --> only drop row where NaN value is present, works on passing arguments.

    3. Fill Data
        Fill by Mean, Median or Mode value
        - Continuous Data --> Mean/Median value
        - Discrete / Categorical Data --> Mode value

        - or fill by custom records if neither Mean, Median nor Mode value satisfy to fill.

        - Skewness check --> This gives the value where Mean and Median actually be at same point of value.
            - This help to choose either Mean or Median Value to fill data
            - If skewness value lies between 0 - 0.5 --> Mean value
            - If skewness value is either less than 0 or more than 0.5 --> Median value
            - Check Skewness of column --> skew()

        - fill data --> fillna()

    4. Remove unwanted Data --> remove like unwanted spaces
        - remove whitespaces (only for object/string) --> strip(), lstrip() or rstrip()
        - remove unwanted texts/characters --> replace()

    5. Convert Datatypes (Typecasting)
        - e.g., when data is in csv format then datatime column data always be in text format. So, we need to convert this datatype to datetime format for further analysis.

# 2. Drop Data
    - dropna() --> only remove Null values
    - drop() --> will remove entire row or column either contain null value or not null value.

```python
# will drop all the rows where Null value is present.
df.dropna(inplace=True) # Permanent change
```

```python
# default
df.dropna()

# For axis:
# 0 -> row
# 1 = column

# For how:
# any -> OR condition
# all -> AND condition

# ROW
# Temporary change, row wise drop, if all column of that row contain Null value
df.dropna(inplace=False, axis=0, how='all')

# Permanent change, row wise drop, if all column of that row contain Null value
df.dropna(inplace=True, axis=0, how='all')

# COLUMN
# Temporary change, column wise drop, if any value contain Null
df.dropna(inplace=False, axis=1, how='any')

# Permanent change, column wise drop, if any value contain Null
df.dropna(inplace=True, axis=1, how='any')
```

# 3. Fill Data

```python
df.isnull().sum()
```

```python
# here 3 columns have Null values
```

```python
df['COMMISSION_PCT'].skew()
```

```python
# here, skew value is 0.09661736135041571 which lies between 0-0.5.
# So, we use Mean value of this column to fill Null values in this column.
```

```python
comm_mean = round(df['COMMISSION_PCT'].mean(), 2)
comm_mean
```

```python
# fill with Mean value
# choose any method

 # This will directly replace the null value with mean with permanent change
df['COMMISSION_PCT'].fillna(comm_mean, inplace=True)

 # This will create new column if the provided column is not found
df['COMMISSION_PCT'] = df['COMMISSION_PCT'].fillna(comm_mean, inplace=True)

 # This will directly replace the null value with mean with permanent change
df['COMMISSION_PCT'] = df['COMMISSION_PCT'].fillna(comm_mean)

 # Passing DICT value
df.fillna({'COMMISSION_PCT': comm_mean}, inplace=True)
```

```python
# There are Discrete Value. So we need Mode of these
mgr_mode = df['MANAGER_ID'].mode()
dept_mode = df['DEPARTMENT_ID'].mode()

print(mgr_mode)
print(dept_mode)
```

```python
# All 3 columns at once
df.fillna({'COMMISSION_PCT': comm_mean}, inplace=True)
df['MANAGER_ID'] = df['MANAGER_ID'].fillna(mgr_mode)
df['DEPARTMENT_ID'] = df['DEPARTMENT_ID'].fillna(dept_mode)
```

```python
df.isnull().sum()
```

```python
# here still giving 1 value because sometimes mode can be more than once. So program get confuse.
# That's why we must select index 0.

comm_mean = round(df['COMMISSION_PCT'].mean(), 2)
mgr_mode = df['MANAGER_ID'].mode()[0]
dept_mode = df['DEPARTMENT_ID'].mode()[0]

df.fillna({'COMMISSION_PCT': comm_mean}, inplace=True)
df['MANAGER_ID'] = df['MANAGER_ID'].fillna(mgr_mode)
df['DEPARTMENT_ID'] = df['DEPARTMENT_ID'].fillna(dept_mode)
```
