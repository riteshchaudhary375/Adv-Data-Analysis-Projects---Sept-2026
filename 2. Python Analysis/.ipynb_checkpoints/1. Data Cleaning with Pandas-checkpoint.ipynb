## Data Cleaning Steps:
    1. Remove duplicate row data
        - check duplicate data --> duplicated()
        - remove duplicate data --> drop_duplicates()

    2. Drop Data
        - If columns contain at least 70-75% Null values then Drop that column.
        - Drop Column --> drop() --> pass column name to drop entire column or pass row index to drop entire row
        - dropna() --> only drop row where NaN value is present, works on passing arguments.

    3. Fill Data
        Fill by Mean, Median or Mode value
        - Continuous Data --> Mean/Median value
        - Discrete / Categorical Data --> Mode value

        - or fill by custom records if neither Mean, Median nor Mode value satisfy to fill.

        - Skewness check --> This gives the value where Mean and Median actually be at same point of value.
            - This help to choose either Mean or Median Value to fill data
            - If skewness value lies between 0 - 0.5 --> Mean value
            - If skewness value is either less than 0 or more than 0.5 --> Median value
            - Check Skewness of column --> skew()

        - fill data --> fillna()

    4. Remove unwanted Data --> remove like unwanted spaces
        - remove whitespaces (only for object/string) --> strip(), lstrip() or rstrip()
        - remove unwanted texts/characters --> replace()

    5. Convert Datatypes (Typecasting)
        - e.g., when data is in csv format then datatime column data always be in text format. So, we need to convert this datatype to datetime format for further analysis.
